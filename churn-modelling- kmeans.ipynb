```python
# ============================================================
# 1. IMPORTAÇÃO DAS BIBLIOTECAS
# ============================================================

import pandas as pd

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 2. VARIÁVEIS UTILIZADAS NO K-MEANS
# ============================================================

variaveis_numericas = [
    'PontuacaoCredito',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado'
]


# ============================================================
# 3. SEPARAÇÃO DOS CLIENTES QUE CANCELARAM
# ============================================================

df_treino_cancelamento = df_treino[df_treino['Saiu'] == 1].copy()


# ============================================================
# 4. PADRONIZAÇÃO DOS DADOS PARA O K-MEANS
# ============================================================

X_cancelamento = df_treino_cancelamento[variaveis_numericas]

scaler = StandardScaler()

X_cancelamento_scaled = scaler.fit_transform(X_cancelamento)


# ============================================================
# 5. CRIAÇÃO DO K-MEANS
# ============================================================

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X_cancelamento_scaled)


# ============================================================
# 6. CRIAÇÃO DOS CLUSTERS NOS CLIENTES QUE CANCELARAM
# ============================================================

df_treino_cancelamento['Cluster'] = kmeans.predict(
    X_cancelamento_scaled
)


# ============================================================
# 7. DISTÂNCIA DE CADA CLIENTE AO CENTROIDE
# ============================================================

distancias_cancelamento = kmeans.transform(
    X_cancelamento_scaled
)

df_treino_cancelamento['Distancia_Centroide'] = [
    distancias_cancelamento[i, cluster]
    for i, cluster in enumerate(
        df_treino_cancelamento['Cluster']
    )
]


# ============================================================
# 8. APLICAR OS CLUSTERS A TODOS OS CLIENTES
# ============================================================
#
# Aqui não estamos simplesmente copiando os clusters.
#
# O K-Means foi treinado com os clientes que cancelaram.
# Agora verificamos com qual perfil de cancelamento cada
# cliente do dataset completo mais se parece.
# ============================================================

X_treino = df_treino[variaveis_numericas]

X_treino_scaled = scaler.transform(X_treino)


# Identifica o perfil de cancelamento mais próximo
df_treino['Cluster'] = kmeans.predict(
    X_treino_scaled
)


# Calcula a distância até os centroides
distancias = kmeans.transform(
    X_treino_scaled
)


# Guarda a distância até o centroide do cluster atribuído
df_treino['Distancia_Centroide'] = [
    distancias[i, cluster]
    for i, cluster in enumerate(
        df_treino['Cluster']
    )
]


# ============================================================
# 9. VISUALIZAÇÃO DOS PERFIS DOS CLUSTERS
# ============================================================

resumo_clusters = (
    df_treino_cancelamento
    .groupby('Cluster')[variaveis_numericas]
    .mean()
)

quantidade_clusters = (
    df_treino_cancelamento['Cluster']
    .value_counts()
    .sort_index()
)

resumo_clusters['Quantidade'] = quantidade_clusters

resumo_clusters['Representatividade'] = (
    resumo_clusters['Quantidade']
    / len(df_treino_cancelamento)
    * 100
)

print('\n==============================')
print('PERFIL DOS CLUSTERS')
print('==============================')

print(resumo_clusters)


# ============================================================
# 10. DEFINIÇÃO DAS VARIÁVEIS DA ÁRVORE DE DECISÃO
# ============================================================

variaveis_numericas_modelo = [
    'PontuacaoCredito',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado',
    'Cluster',
    'Distancia_Centroide'
]

variaveis_categoricas = [
    'Geografia',
    'Genero'
]

variaveis_modelo = (
    variaveis_numericas_modelo
    + variaveis_categoricas
)


# ============================================================
# 11. DEFINIÇÃO DE X E Y
# ============================================================

X = df_treino[variaveis_modelo]

y = df_treino['Saiu']


# ============================================================
# 12. DIVISÃO ENTRE TREINO E TESTE
# ============================================================

X_treino_modelo, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 13. PRÉ-PROCESSAMENTO DAS VARIÁVEIS CATEGÓRICAS
# ============================================================

preprocessador = ColumnTransformer(
    transformers=[
        (
            'categoricas',
            OneHotEncoder(handle_unknown='ignore'),
            variaveis_categoricas
        )
    ],
    remainder='passthrough'
)


# ============================================================
# 14. CRIAÇÃO DA ÁRVORE DE DECISÃO
# ============================================================

arvore_decisao = DecisionTreeClassifier(
    random_state=42,
    class_weight='balanced'
)


# ============================================================
# 15. PIPELINE COMPLETO
# ============================================================

modelo = Pipeline(
    steps=[
        ('preprocessador', preprocessador),
        ('arvore', arvore_decisao)
    ]
)


# ============================================================
# 16. TREINAMENTO DO MODELO
# ============================================================

modelo.fit(
    X_treino_modelo,
    y_treino
)


# ============================================================
# 17. PREVISÕES NO CONJUNTO DE TESTE
# ============================================================

y_pred = modelo.predict(X_teste)

y_prob = modelo.predict_proba(X_teste)[:, 1]


# ============================================================
# 18. AVALIAÇÃO DO MODELO
# ============================================================

acuracia = accuracy_score(
    y_teste,
    y_pred
)

print('\n==============================')
print('AVALIAÇÃO DA ÁRVORE DE DECISÃO')
print('==============================')

print(f'\nAcurácia: {acuracia:.2%}')

print('\nRelatório de Classificação:')
print(
    classification_report(
        y_teste,
        y_pred
    )
)


# ============================================================
# 19. MATRIZ DE CONFUSÃO
# ============================================================

matriz = confusion_matrix(
    y_teste,
    y_pred
)

print('\nMatriz de Confusão:')
print(matriz)


# ============================================================
# 20. PREVISÃO PARA TODOS OS CLIENTES
# ============================================================

df_treino['Probabilidade_Cancelamento'] = (
    modelo.predict_proba(
        df_treino[variaveis_modelo]
    )[:, 1]
)


df_treino['Previsao_Cancelamento'] = (
    modelo.predict(
        df_treino[variaveis_modelo]
    )
)


# ============================================================
# 21. VISUALIZAÇÃO DOS RESULTADOS
# ============================================================

colunas_resultado = [
    'IdCliente',
    'PontuacaoCredito',
    'Geografia',
    'Genero',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado',
    'Saiu',
    'Cluster',
    'Distancia_Centroide',
    'Probabilidade_Cancelamento',
    'Previsao_Cancelamento'
]

print('\n==============================')
print('RESULTADO FINAL')
print('==============================')

print(
    df_treino[colunas_resultado].head(20)
)
```
